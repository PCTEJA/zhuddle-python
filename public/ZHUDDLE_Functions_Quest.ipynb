````python
SOURCE_URL = 'https://www.py4e.com/html3/04-functions'
MCQS = [{'id': 'M01', 'title': 'Define now, call later', 'level': 'Easy', 'source': 'Adding new functions; Exercise 4', 'prompt': '### M01 | Define now, call later | Easy | 2 XP\n\nWhat does `def` do?\n\n- **A.** Immediately runs the function body.\n- **B.** Defines a function whose body runs when called.\n- **C.** Displays the return value.\n- **D.** Imports a module.\n\nSource checkpoint: Adding new functions; Exercise 4.', 'answer': 'B'}, {'id': 'M02', 'title': 'Follow the calls', 'level': 'Easy', 'source': 'Exercise 5', 'prompt': '### M02 | Follow the calls | Easy | 2 XP\n\nWhat is the output, in order?\n\n```python\ndef fred():\n    print("Zap")\n\ndef jane():\n    print("ABC")\n\njane()\nfred()\njane()\n```\n\n- **A.** Zap, ABC, Zap\n- **B.** ABC, ABC, Zap\n- **C.** ABC, Zap, ABC\n- **D.** No output\n\nSource checkpoint: Exercise 5.', 'answer': 'C'}, {'id': 'M03', 'title': 'The value supplied', 'level': 'Easy', 'source': 'Function calls', 'prompt': '### M03 | The value supplied | Easy | 2 XP\n\nIn `type(32)`, which item is the argument?\n\n- **A.** type\n- **B.** int\n- **C.** 32\n- **D.** return\n\nSource checkpoint: Function calls.', 'answer': 'C'}, {'id': 'M04', 'title': 'Conversion check', 'level': 'Easy', 'source': 'Type conversion functions', 'prompt': '### M04 | Conversion check | Easy | 2 XP\n\nWhat does `int(-2.3)` return?\n\n- **A.** -3\n- **B.** -2\n- **C.** 2\n- **D.** -2.3\n\nSource checkpoint: Type conversion functions.', 'answer': 'B'}, {'id': 'M05', 'title': 'A module call', 'level': 'Easy', 'source': 'Math functions', 'prompt': '### M05 | A module call | Easy | 2 XP\n\nAfter `import math`, which expression calls the square-root function?\n\n- **A.** math.sqrt(2)\n- **B.** math: sqrt(2)\n- **C.** import.sqrt(2)\n- **D.** sqrt.math(2)\n\nSource checkpoint: Math functions.', 'answer': 'A'}, {'id': 'M06', 'title': 'Random range', 'level': 'Easy', 'source': 'Random numbers; Exercise 1', 'prompt': '### M06 | Random range | Easy | 2 XP\n\nWhich interval describes `random.random()`?\n\n- **A.** 0.0 through 1.0, both included\n- **B.** Greater than 0.0 through 1.0, including 1.0\n- **C.** Only the integers 0 and 1\n- **D.** At least 0.0 and less than 1.0\n\nSource checkpoint: Random numbers; Exercise 1.', 'answer': 'D'}, {'id': 'M07', 'title': 'Two names, one value', 'level': 'Easy', 'source': 'Parameters and arguments', 'prompt': '### M07 | Two names, one value | Easy | 2 XP\n\nFor `def print_twice(bruce):`, what is `bruce`?\n\n- **A.** A parameter name\n- **B.** A required name in the calling code\n- **C.** A module\n- **D.** The function name\n\nSource checkpoint: Parameters and arguments.', 'answer': 'A'}, {'id': 'M08', 'title': 'An action without a result', 'level': 'Easy', 'source': 'Fruitful functions and void functions', 'prompt': '### M08 | An action without a result | Easy | 2 XP\n\nA function prints text but has no `return` statement. What is assigned by `result = that_function()`?\n\n- **A.** The printed text\n- **B.** The string "None"\n- **C.** None\n- **D.** 0\n\nSource checkpoint: Fruitful functions and void functions.', 'answer': 'C'}, {'id': 'M09', 'title': 'Definition order', 'level': 'Medium', 'source': 'Definitions and uses; Exercise 3', 'prompt': '### M09 | Definition order | Medium | 2 XP\n\nA function is defined first, a helper it calls is defined second, and the first function is called last. What happens?\n\n- **A.** It always fails because the helper appears second.\n- **B.** It can run because both definitions exist before the call.\n- **C.** Both bodies run during definition.\n- **D.** Only the helper runs.\n\nSource checkpoint: Definitions and uses; Exercise 3.', 'answer': 'B'}, {'id': 'M10', 'title': 'Keep the result', 'level': 'Medium', 'source': 'Fruitful functions and void functions', 'prompt': '### M10 | Keep the result | Medium | 2 XP\n\nA script contains only `import math` and `math.sqrt(5)`. What happens to that result?\n\n- **A.** It is automatically printed.\n- **B.** It is stored as x.\n- **C.** It is returned from the whole script.\n- **D.** It is computed but is neither stored nor displayed.\n\nSource checkpoint: Fruitful functions and void functions.', 'answer': 'D'}]
QUESTS = {'C01': {'title': 'Toolbox warm-up', 'level': 'Easy', 'source': 'Built-in functions; Type conversion functions', 'prompt': '## C01 | Toolbox warm-up\n\n**Easy | 10 XP | Five checks worth 2 points each**\n\nReplace each `None` with one function call. Use `len`, `max`, and `min` on `message`, `int` on `"32"`, and `float` on `"3.14159"`. Keep the variable names. Each correctly computed variable earns 2 points. Do not type the final values directly.\n\n**Hint:** A call has a function name followed by an argument in parentheses.\n\n**Source checkpoint:** Built-in functions; Type conversion functions.', 'starter': 'message = "Hello world"\ncharacter_count = None\nlargest_character = None\nsmallest_character = None\nwhole_number = None\ndecimal_number = None'}, 'C02': {'title': 'Random-number lab', 'level': 'Easy', 'source': 'Random numbers; Exercise 1', 'prompt': '## C02 | Random-number lab\n\n**Easy | 10 XP | Five checks worth 2 points each**\n\nComplete the three TODOs, keeping the supplied loop. Import `random`, assign a call to `random.random()` to `x`, and print `x` inside the loop. Run your cell twice and compare the outputs. The checks award 2 points each for the import, ten-iteration loop, random call inside the loop, ten numeric output lines, and values in the correct interval. No particular random sequence is required. The loop is supplied; you do not need to design one.\n\n**Hint:** Use module.function() notation. Leave the indented lines inside the loop.\n\n**Source checkpoint:** Random numbers; Exercise 1.', 'starter': '# TODO: import the module\n\nfor i in range(10):\n    x = None  # TODO: call the function\n    # TODO: print x'}, 'C03': {'title': 'Repair the order', 'level': 'Easy', 'source': 'Definitions and uses; Exercises 2 and 3', 'prompt': "## C03 | Repair the order\n\n**Easy | 10 XP | Five checks worth 2 points each**\n\nThe first line is in the wrong place. Run the starter once to observe the error. Then move `repeat_message()` below both definitions. Keep `repeat_message` defined before `print_message`. The successful program prints `Hello world` twice, on separate lines. This adapts the chapter's two-function example using another string from the chapter. Checks cover both definitions, their requested order, two helper calls, and the final call/output.\n\n**Hint:** Defining a function does not run its body. Both definitions must exist when the final call executes.\n\n**Source checkpoint:** Definitions and uses; Exercises 2 and 3.", 'starter': 'repeat_message()\n\ndef repeat_message():\n    print_message()\n    print_message()\n\ndef print_message():\n    print("Hello world")'}, 'C04': {'title': 'Print twice', 'level': 'Easy', 'source': 'Parameters and arguments; Fruitful functions and void functions', 'prompt': '## C04 | Print twice\n\n**Easy | 10 XP | Five checks worth 2 points each**\n\nReplace the placeholder body with two print statements using `bruce`. Then replace the final `None` with a call to your function using `"Bing"`. Store that call\'s result in `result`. Do not add a return statement or print `result` in this cell. Checks cover the definition, text input, numeric input, an expression input, and the saved result/sample output.\n\n**Hint:** Use the parameter so the function works with different arguments.\n\n**Source checkpoint:** Parameters and arguments; Fruitful functions and void functions.', 'starter': 'def print_twice(bruce):\n    ...  # Replace this line\n\nresult = None  # Replace with your function call'}, 'C05': {'title': 'Return a sum', 'level': 'Easy', 'source': 'Fruitful functions and void functions; addtwo example', 'prompt': '## C05 | Return a sum\n\n**Easy | 10 XP | Five checks worth 2 points each**\n\nComplete `addtwo(a, b)` so it returns the sum of its arguments. Call it with `3` and `5`, store the return value in `x`, and print `x` after the definition. The function itself should not print. Checks cover the definition, three input pairs, and the sample call/output.\n\n**Hint:** A fruitful function sends its computed value back with return.\n\n**Source checkpoint:** Fruitful functions and void functions; addtwo example.', 'starter': 'def addtwo(a, b):\n    return None  # Replace None\n\nx = None  # Replace with your function call\n# Print x here'}, 'C06': {'title': 'Compose the math calls', 'level': 'Medium', 'source': 'Math functions; composition example', 'prompt': "## C06 | Compose the math calls\n\n**Medium | 10 XP | Five checks worth 2 points each**\n\nImport `math`. Replace `None` values to reproduce the chapter's 45-degree example: compute `radians` with `degrees / 360.0 * 2 * math.pi`, compute `height` by calling `math.sin(radians)`, and compute `comparison` as `math.sqrt(2) / 2.0`. Print `height`, then `comparison`, on separate lines. Each setup/calculation/output check is worth 2 points. The formulas are supplied; no trigonometry knowledge is required.\n\n**Hint:** The result of a function call can be part of a larger expression.\n\n**Source checkpoint:** Math functions; composition example.", 'starter': '# Import the module here\ndegrees = 45\nradians = None\nheight = None\ncomparison = None\n# Print the two results here'}, 'C07': {'title': 'Overtime checkpoint', 'level': 'Medium', 'source': 'Exercise 6', 'prompt': '## C07 | Overtime checkpoint\n\n**Medium | 10 XP | Five checks worth 2 points each**\n\nFill only the three `None` placeholders. Return pay at the regular rate for up to 40 hours. For more than 40 hours, pay the first 40 at `rate` and the remaining hours at `1.5 * rate`. Call `computepay(45, 10)` and store the result in `pay`. The supplied print should show `475` or `475.0`. The branching structure is supplied because the chapter refers back to an earlier pay exercise. Checks cover the function, below/exactly/above 40 hours, and a second overtime rate plus the sample output.\n\n**Hint:** Overtime pay is 40 * rate plus (hours - 40) * rate * 1.5.\n\n**Source checkpoint:** Exercise 6.', 'starter': 'def computepay(hours, rate):\n    # Supplied condition: complete only the return expressions.\n    if hours > 40:\n        return None\n    else:\n        return None\n\npay = None  # Call your function with 45 and 10\nprint(pay)'}, 'C08': {'title': 'Grade gate', 'level': 'Medium', 'source': 'Exercise 7', 'prompt': '## C08 | Grade gate\n\n**Medium | 10 XP | Five checks worth 2 points each**\n\nReplace the six empty strings inside `computegrade` with the correct return strings: `"Bad score"` for a score outside 0 through 1, `"A"` for at least 0.9, `"B"` for at least 0.8, `"C"` for at least 0.7, `"D"` for at least 0.6, and `"F"` below 0.6. Keep the supplied conditions and helper unchanged. `grade_text` handles converting text and invalid text such as `"perfect"`; writing exception handling is not part of this assignment. Checks cover the definition, upper boundaries, lower boundaries, invalid numeric scores, and the supplied helper/sample output.\n\n**Hint:** Return quoted strings. The conditions are already in order from highest threshold to lowest.\n\n**Source checkpoint:** Exercise 7.', 'starter': 'def computegrade(score):\n    # Supplied conditions: fill only the six return strings.\n    if score < 0 or score > 1:\n        return ""\n    if score >= 0.9:\n        return ""\n    elif score >= 0.8:\n        return ""\n    elif score >= 0.7:\n        return ""\n    elif score >= 0.6:\n        return ""\n    else:\n        return ""\n\n\n# PROVIDED SUPPORT: do not change; exception handling is not assessed.\ndef grade_text(text):\n    try:\n        number = float(text)\n    except (ValueError, TypeError):\n        return "Bad score"\n    return computegrade(number)\n\nprint(grade_text("0.95"))\nprint(grade_text("perfect"))'}}

import ast
import contextlib
import hashlib
import html
import io
import json
import math
import re
import sys
import uuid
from datetime import datetime, timezone
from pathlib import Path
from IPython.display import HTML, FileLink, display

_answers = {}
_mcq_answers = {}
_student = {}
_notes = ''


def set_student(name, unt_id):
    name, unt_id = str(name).strip(), str(unt_id).strip()
    if not name or not unt_id:
        raise ValueError('Enter both your name and your UNT ID, then rerun this cell.')
    if len(name) > 100 or len(unt_id) > 64 or any(ord(c) < 32 for c in name + unt_id):
        raise ValueError('Use a name up to 100 characters and an ID up to 64, without control characters.')
    if _student and _student != {'name': name, 'unt_id': unt_id}:
        raise ValueError('For a different student, restart the kernel and begin again.')
    _student.update(name=name, unt_id=unt_id)
    print('Welcome, ' + name + '. Your quest has 100 available points.')


@contextlib.contextmanager
def _bounded():
    # A trace budget catches accidental infinite Python loops, not hostile code.
    count = 0
    previous = sys.gettrace()
    def trace(frame, event, arg):
        nonlocal count
        if event == 'line':
            count += 1
            if count > 20000:
                raise RuntimeError('This run exceeded the practice limit. Check for an endless loop.')
        return trace
    sys.settrace(trace)
    try:
        yield
    finally:
        sys.settrace(previous)


def _called(tree, name):
    return any(isinstance(n, ast.Call) and isinstance(n.func, ast.Name) and n.func.id == name
               for n in ast.walk(tree))


def _attribute_called(tree, module, name):
    return any(isinstance(n, ast.Call) and isinstance(n.func, ast.Attribute)
               and isinstance(n.func.value, ast.Name) and n.func.value.id == module
               and n.func.attr == name for n in ast.walk(tree))


def _definition(tree, name, count):
    return any(isinstance(n, ast.FunctionDef) and n.name == name
               and len(n.args.args) == count for n in tree.body)


def _invoke(ns, name, args, expected, printed=None):
    sink = io.StringIO()
    with _bounded(), contextlib.redirect_stdout(sink):
        value = ns[name](*args)
    same = (type(value) in (int, float) and math.isclose(value, expected, abs_tol=1e-9)) if type(expected) in (int, float) else value == expected
    return same and (printed is None or sink.getvalue() == printed)


def _evaluate(qid, source):
    ns = {'__name__': '__student__'}
    output = io.StringIO()
    error = None
    tree = None
    try:
        tree = ast.parse(source)
        with _bounded(), contextlib.redirect_stdout(output), contextlib.redirect_stderr(output):
            exec(compile(tree, '<' + qid + '>', 'exec'), ns)
    except BaseException as exc:
        if isinstance(exc, KeyboardInterrupt):
            error = 'KeyboardInterrupt: run interrupted; edit and retry.'
        else:
            error = type(exc).__name__ + ': ' + str(exc)
    checks = []
    def check(label, predicate):
        try:
            passed = error is None and bool(predicate())
        except BaseException:
            passed = False
        checks.append({'label': label, 'earned': 2 if passed else 0, 'possible': 2})
    if qid == 'C01':
        check('Use len for the character count.', lambda: ns.get('character_count') == 11 and _called(tree, 'len'))
        check('Use max for the largest character.', lambda: ns.get('largest_character') == 'w' and _called(tree, 'max'))
        check('Use min for the smallest character.', lambda: ns.get('smallest_character') == ' ' and _called(tree, 'min'))
        check('Use int to convert the text value.', lambda: type(ns.get('whole_number')) is int and ns['whole_number'] == 32 and _called(tree, 'int'))
        check('Use float to convert the decimal text.', lambda: type(ns.get('decimal_number')) is float and ns['decimal_number'] == 3.14159 and _called(tree, 'float'))
    elif qid == 'C02':
        lines = output.getvalue().splitlines()
        check('Import the random module.', lambda: any(isinstance(n, ast.Import) and any(a.name == 'random' for a in n.names) for n in tree.body))
        check('Use the supplied for loop and range(10).', lambda: any(isinstance(n, ast.For) and isinstance(n.iter, ast.Call) and isinstance(n.iter.func, ast.Name) and n.iter.func.id == 'range' and len(n.iter.args) == 1 and isinstance(n.iter.args[0], ast.Constant) and n.iter.args[0].value == 10 for n in ast.walk(tree)))
        check('Call random.random inside the loop.', lambda: any(isinstance(n, ast.For) and _attribute_called(n, 'random', 'random') for n in ast.walk(tree)))
        check('Print exactly ten numeric lines.', lambda: len(lines) == 10 and all(math.isfinite(float(v)) for v in lines))
        check('All ten values are at least 0 and below 1.', lambda: len(lines) == 10 and all(0 <= float(v) < 1 for v in lines))
    elif qid == 'C03':
        check('Define repeat_message with no parameters.', lambda: _definition(tree, 'repeat_message', 0))
        check('Define print_message with no parameters.', lambda: _definition(tree, 'print_message', 0))
        check('Define repeat_message before print_message.', lambda: [n.name for n in tree.body if isinstance(n, ast.FunctionDef)] == ['repeat_message', 'print_message'])
        check('Call print_message twice inside repeat_message.', lambda: sum(isinstance(n, ast.Call) and isinstance(n.func, ast.Name) and n.func.id == 'print_message' for f in tree.body if isinstance(f, ast.FunctionDef) and f.name == 'repeat_message' for n in ast.walk(f)) == 2)
        check('Call repeat_message after the definitions; print two lines.', lambda: isinstance(tree.body[-1], ast.Expr) and _called(tree.body[-1], 'repeat_message') and output.getvalue() == 'Hello world\nHello world\n')
    elif qid == 'C04':
        check('Define print_twice with one parameter.', lambda: _definition(tree, 'print_twice', 1))
        check('Print the text argument twice and return None.', lambda: _invoke(ns, 'print_twice', ('Spam',), None, 'Spam\nSpam\n'))
        check('Print a numeric argument twice.', lambda: _invoke(ns, 'print_twice', (17,), None, '17\n17\n'))
        check('Print an expression argument twice.', lambda: _invoke(ns, 'print_twice', ('Spam ' * 4,), None, ('Spam ' * 4 + '\n') * 2))
        check('Call your function with Bing and store its None result.', lambda: 'result' in ns and ns['result'] is None and output.getvalue() == 'Bing\nBing\n' and _called(tree, 'print_twice'))
    elif qid == 'C05':
        check('Define addtwo with two parameters.', lambda: _definition(tree, 'addtwo', 2))
        check('Return the sum of 3 and 5.', lambda: _invoke(ns, 'addtwo', (3, 5), 8, ''))
        check('Return the sum of 0 and 7.', lambda: _invoke(ns, 'addtwo', (0, 7), 7, ''))
        check('Return the sum of -2 and 5.', lambda: _invoke(ns, 'addtwo', (-2, 5), 3, ''))
        check('Store addtwo(3, 5) in x and print x.', lambda: ns.get('x') == 8 and output.getvalue() == '8\n' and _called(tree, 'addtwo'))
    elif qid == 'C06':
        check('Import math and keep degrees equal to 45.', lambda: ns.get('degrees') == 45 and any(isinstance(n, ast.Import) and any(a.name == 'math' for a in n.names) for n in tree.body))
        check('Compute radians from degrees using math.pi.', lambda: math.isclose(ns['radians'], math.pi / 4) and any(isinstance(n, ast.Attribute) and n.attr == 'pi' for n in ast.walk(tree)))
        check('Compute height with math.sin(radians).', lambda: math.isclose(ns['height'], math.sin(math.pi / 4)) and _attribute_called(tree, 'math', 'sin'))
        check('Compute comparison with math.sqrt(2) divided by 2.', lambda: math.isclose(ns['comparison'], math.sqrt(2) / 2) and _attribute_called(tree, 'math', 'sqrt'))
        check('Print height then comparison.', lambda: len(output.getvalue().splitlines()) == 2 and all(math.isclose(float(v), math.sqrt(2) / 2) for v in output.getvalue().splitlines()))
    elif qid == 'C07':
        check('Define computepay with hours and rate parameters.', lambda: _definition(tree, 'computepay', 2))
        check('Return 300 for 30 hours at 10.', lambda: _invoke(ns, 'computepay', (30, 10), 300, ''))
        check('Return 400 for exactly 40 hours at 10.', lambda: _invoke(ns, 'computepay', (40, 10), 400, ''))
        check('Return 475 for 45 hours at 10.', lambda: _invoke(ns, 'computepay', (45, 10), 475, ''))
        check('Handle another overtime rate and print the sample pay.', lambda: _invoke(ns, 'computepay', (42, 20), 860, '') and output.getvalue().strip() in ('475', '475.0'))
    elif qid == 'C08':
        check('Define computegrade with one parameter.', lambda: _definition(tree, 'computegrade', 1))
        check('Return A and B at the correct boundaries.', lambda: all(_invoke(ns, 'computegrade', (v,), grade, '') for v, grade in [(1.0,'A'),(.95,'A'),(.9,'A'),(.899,'B'),(.8,'B')]))
        check('Return C, D and F at the correct boundaries.', lambda: all(_invoke(ns, 'computegrade', (v,), grade, '') for v, grade in [(.799,'C'),(.75,'C'),(.7,'C'),(.699,'D'),(.6,'D'),(.599,'F'),(.5,'F'),(0,'F')]))
        check('Reject numeric scores outside 0 through 1.', lambda: all(_invoke(ns, 'computegrade', (v,), 'Bad score', '') for v in (-.1, 10.0)))
        check('Use the supplied text-input helper for numeric and invalid text.', lambda: all(_invoke(ns, 'grade_text', (v,), grade, '') for v, grade in [('0.95','A'),('0.75','C'),('0.5','F'),('perfect','Bad score'),('10.0','Bad score')]) and output.getvalue() == 'A\nBad score\n')
    return {'source': source, 'stdout': output.getvalue(), 'error': error,
            'checks': checks, 'earned': sum(c['earned'] for c in checks), 'possible': 10}


def answer(line, cell):
    qid = line.strip()
    if qid not in QUESTS:
        raise ValueError('Keep the first line of each answer cell unchanged.')
    attempts = _answers.get(qid, {}).get('attempts', 0) + 1
    result = _evaluate(qid, cell)
    result['attempts'] = attempts
    _answers[qid] = result
    if result['stdout']:
        print(result['stdout'], end='')
    if result['error']:
        print(result['error'])
    print('\n' + qid + ': ' + str(result['earned']) + '/10 points | attempt ' + str(attempts))
    for row in result['checks']:
        print(('PASS  ' if row['earned'] else 'RETRY ') + row['label'] + ' (' + str(row['earned']) + '/2)')
    show_progress()


get_ipython().register_magic_function(answer, 'cell', 'answer')


def submit_mcqs(*choices):
    if len(choices) != len(MCQS):
        raise ValueError('Keep all ten answer variables in the supplied call.')
    for question, value in zip(MCQS, choices):
        choice = str(value).strip().upper()
        correct = choice == question['answer']
        _mcq_answers[question['id']] = {'choice': choice, 'earned': 2 if correct else 0}
        status = 'PASS' if correct else ('UNANSWERED' if not choice else 'RETRY')
        print(question['id'] + ': ' + status + ' (' + str(2 if correct else 0) + '/2)')
    show_progress()


def _total():
    return sum(v['earned'] for v in _answers.values()) + sum(v['earned'] for v in _mcq_answers.values())


def _badge(score):
    if score == 100:
        return 'Function Champion'
    if score >= 75:
        return 'Function Builder'
    if score >= 50:
        return 'Function Explorer'
    return 'Function Starter'


def show_progress():
    score = _total()
    completed = sum(v['earned'] == 10 for v in _answers.values())
    display(HTML('<div style="border-left:5px solid #008568;padding:12px;background:#f2f7f5;color:#16372c">'
                 '<b>' + str(score) + '/100 XP | ' + _badge(score) + '</b><br>'
                 + str(completed) + '/8 coding missions mastered'
                 '<div style="height:10px;background:#d6dfdc;margin-top:8px"><div style="height:10px;width:'
                 + str(score) + '%;background:#008568"></div></div></div>'))


def save_notes(text):
    global _notes
    _notes = str(text)
    print('Reflection saved. This reflection carries no points.')


def _rows():
    result = []
    for q in MCQS:
        a = _mcq_answers.get(q['id'], {'choice':'', 'earned':0})
        feedback = ('Correct.' if a['earned'] else 'Review: ' + q['source'])
        result.append({'id':q['id'], 'title':q['title'], 'level':q['level'],
                       'earned':a['earned'], 'possible':2,
                       'feedback':'Selected: ' + (a['choice'] or '(unanswered)') + '. ' + feedback})
    for qid, q in QUESTS.items():
        a = _answers.get(qid)
        feedback = 'Not run. No points awarded.' if a is None else '; '.join(
            ('PASS: ' if c['earned'] else 'RETRY: ') + c['label'] for c in a['checks'])
        if a and a['error']:
            feedback = a['error'] + '. ' + feedback
        result.append({'id':qid, 'title':q['title'], 'level':q['level'],
                       'earned':a['earned'] if a else 0, 'possible':10, 'feedback':feedback})
    return result


def _md(text):
    return {'cell_type':'markdown', 'id':uuid.uuid4().hex[:8], 'metadata':{}, 'source':text}


def _code(source, output='', count=None):
    return {'cell_type':'code', 'id':uuid.uuid4().hex[:8], 'metadata':{}, 'source':source,
            'execution_count':count, 'outputs':([{'output_type':'stream','name':'stdout','text':output}] if output else [])}


def _submission(receipt):
    cells = [_md('# ZHUDDLE | Functions Quest\n\nStudent submission\n\n'
                 + '**Name:** ' + _student['name'] + '\n\n**UNT ID:** ' + _student['unt_id']
                 + '\n\n**Practice grade:** ' + str(_total()) + '/100\n\n**Badge:** ' + _badge(_total())
                 + '\n\nReceipt: ' + receipt['id'] + '\n\nUTC: ' + receipt['created_utc']
                 + '\n\nAssignment: ' + SOURCE_URL
                 + '\n\nThis file contains the latest executed answer for every assignment question, captured output, feedback, and reflection. The original answer notebook remains your editable workspace. Scores are local self-check results pending instructor review.')]
    for q in MCQS:
        cells.append(_md(q['prompt']))
        a = _mcq_answers.get(q['id'], {'choice':'','earned':0})
        cells.append(_code(q['id'].lower() + ' = ' + repr(a['choice'])))
        cells.append(_md('Score: ' + str(a['earned']) + '/2'))
    for qid, q in QUESTS.items():
        a = _answers.get(qid)
        cells.append(_md(q['prompt']))
        cells.append(_code(a['source'] if a else q['starter'], a['stdout'] if a else ''))
        row = next(r for r in receipt['results'] if r['id'] == qid)
        cells.append(_md('Score: ' + str(row['earned']) + '/10\n\n' + row['feedback']))
    cells.append(_md('## Reflection\n\n' + (_notes or '(No reflection entered.)')))
    return {'nbformat':4,'nbformat_minor':5, 'metadata':{
        'kernelspec':{'display_name':'Python 3','language':'python','name':'python3'},
        'language_info':{'name':'python','version':sys.version.split()[0]}, 'zhuddle':receipt}, 'cells':cells}


def _pdf(path, receipt):
    from reportlab.lib import colors
    from reportlab.lib.styles import getSampleStyleSheet, ParagraphStyle
    from reportlab.lib.enums import TA_LEFT
    from reportlab.lib.pagesizes import letter
    from reportlab.platypus import SimpleDocTemplate, Paragraph, Spacer, Table, TableStyle, PageBreak, KeepTogether
    from reportlab.pdfbase import pdfmetrics
    from reportlab.pdfbase.ttfonts import TTFont
    import reportlab
    font_path = Path(reportlab.__file__).parent / 'fonts' / 'Vera.ttf'
    pdfmetrics.registerFont(TTFont('QuestFont', str(font_path)))
    styles = getSampleStyleSheet()
    for style in styles.byName.values():
        style.fontName = 'QuestFont'
    styles.add(ParagraphStyle(name='QuestSmall',fontName='QuestFont',fontSize=8.5,leading=12,spaceAfter=5,wordWrap='CJK'))
    styles['Title'].textColor = colors.HexColor('#00765e')
    styles['BodyText'].leading = 14
    esc = lambda value: html.escape(str(value))
    p = lambda text, style='BodyText': Paragraph(esc(text), styles[style])
    story = [p('ZHUDDLE | Functions Quest', 'Title'), p('GRADE REPORT - LOCAL SELF-CHECK', 'Heading2'),
             p('Student: ' + _student['name']), p('UNT ID: ' + _student['unt_id']),
             Spacer(1,12), p(str(_total()) + ' / 100 points', 'Title'), p(_badge(_total()), 'Heading2'),
             p('MCQs: ' + str(sum(v['earned'] for v in _mcq_answers.values())) + '/20 | Coding: ' + str(sum(v['earned'] for v in _answers.values())) + '/80'),
             p('Latest executed answers count. Retries do not reduce points. Unanswered items earn zero.'),
             p('These results require instructor review; this is not an official UNT transcript.'), Spacer(1,12)]
    table = [[p('Question','QuestSmall'),p('Level','QuestSmall'),p('Points','QuestSmall')]]
    for row in receipt['results']:
        table.append([p(row['id'] + '  ' + row['title'],'QuestSmall'),p(row['level'],'QuestSmall'),p(str(row['earned']) + '/' + str(row['possible']),'QuestSmall')])
    t = Table(table, colWidths=[340,80,80], repeatRows=1, hAlign='LEFT')
    t.setStyle(TableStyle([('BACKGROUND',(0,0),(-1,0),colors.HexColor('#d9eee6')),
                          ('ROWBACKGROUNDS',(0,1),(-1,-1),[colors.white,colors.HexColor('#f5f7f6')]),
                          ('VALIGN',(0,0),(-1,-1),'TOP'),('BOTTOMPADDING',(0,0),(-1,-1),4),('TOPPADDING',(0,0),(-1,-1),4)]))
    story.extend([t, PageBreak(), p('Question Feedback', 'Heading1')])
    for row in receipt['results']:
        story.append(KeepTogether([p(row['id'] + ' | ' + row['title'] + ' | ' + str(row['earned']) + '/' + str(row['possible']),'Heading3'),p(row['feedback'],'QuestSmall')]))
    story.extend([Spacer(1,12),p('Submission Receipt','Heading2'),p('Receipt: ' + receipt['id'],'QuestSmall'),
                  p('Created (UTC): ' + receipt['created_utc'],'QuestSmall'),
                  p('Answer SHA-256: ' + receipt['answer_sha256'],'QuestSmall'),
                  p('Source: ' + SOURCE_URL,'QuestSmall'),
                  p('Adapted from Charles R. Severance, Python for Everybody, Functions. CC BY 4.0. Original question wording and scaffolding added for this assignment.','QuestSmall')])
    def footer(canvas, doc):
        canvas.setFont('QuestFont',8)
        canvas.setFillColor(colors.HexColor('#52645e'))
        canvas.drawString(48,28,'ZHUDDLE | Functions Quest | ' + receipt['id'][:8])
        canvas.drawRightString(564,28,'Page ' + str(doc.page))
    SimpleDocTemplate(str(path),pagesize=letter,rightMargin=48,leftMargin=48,topMargin=40,bottomMargin=48,
                      title='ZHUDDLE Functions Quest Grade Report',author='ZHUDDLE').build(story,onFirstPage=footer,onLaterPages=footer)


def finalize(output_directory='zhuddle_submissions'):
    if not _student:
        raise ValueError('Run the student name and UNT ID cell before exporting.')
    try:
        import reportlab
    except ImportError as exc:
        raise ImportError('PDF export needs ReportLab. Run %pip install reportlab in a new code cell, then rerun this cell. Your answers are still in memory.') from exc
    receipt = {'id':uuid.uuid4().hex,'created_utc':datetime.now(timezone.utc).isoformat(),
               'student':dict(_student),'score':_total(),'possible':100,'badge':_badge(_total()),
               'results':_rows(),'version':'1.0'}
    evidence = {'code':{k:v['source'] for k,v in _answers.items()},'mcqs':_mcq_answers,'reflection':_notes}
    receipt['answer_sha256'] = hashlib.sha256(json.dumps(evidence,sort_keys=True,ensure_ascii=False).encode()).hexdigest()
    safe_id = re.sub(r'[^A-Za-z0-9_-]', '_', _student['unt_id'])[:64] or 'student'
    target = Path(output_directory) / (safe_id + '_' + receipt['id'][:8])
    target.mkdir(parents=True,exist_ok=False)
    nb_path = target / 'ZHUDDLE_Student_Submission.ipynb'
    pdf_path = target / 'ZHUDDLE_Grade_Report.pdf'
    nb_path.write_text(json.dumps(_submission(receipt),ensure_ascii=False,indent=1),encoding='utf-8')
    _pdf(pdf_path, receipt)
    print('Export complete: ' + str(_total()) + '/100 | ' + _badge(_total()))
    print('Submit BOTH files. Each export gets a new folder; earlier exports remain available.')
    display(FileLink(str(nb_path)))
    display(FileLink(str(pdf_path)))
    print('In Colab, find the same folder in the Files sidebar and download both files.')
    return {'notebook':str(nb_path), 'pdf':str(pdf_path), 'receipt':receipt}


print('Quest engine ready. Continue to student sign-in. This supplied cell is not assessed.')
````
